```python

# ================================================================
# DRINKING WATER POTABILITY PREDICTION USING MACHINE LEARNING
# End-to-End Predictive Analytics | KDD Process
# Models: Logistic Regression, Decision Tree, Random Forest, XGBoost
# ================================================================

# KDD STAGES:
# 1. Problem Understanding
# 2. Data Selection
# 3. Data Understanding / EDA
# 4. Data Cleaning
# 5. Data Transformation
# 6. Feature Selection
# 7. Data Mining / Model Development
# 8. Model Evaluation
# 9. Interpretation and Knowledge Presentation

# ================================================================
# 1. IMPORT LIBRARIES
# ================================================================

import os
import sys
import warnings
import subprocess
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import joblib

from IPython.display import display, Markdown
from sklearn.model_selection import (
    train_test_split, StratifiedKFold, cross_val_score
)
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.feature_selection import mutual_info_classif
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    roc_auc_score, confusion_matrix, classification_report,
    ConfusionMatrixDisplay, RocCurveDisplay
)

warnings.filterwarnings("ignore")
sns.set_theme(style="whitegrid")
RANDOM_STATE = 42

# Install XGBoost if it is not already available.
try:
    from xgboost import XGBClassifier
except ImportError:
    print("Installing XGBoost...")
    subprocess.check_call(
        [sys.executable, "-m", "pip", "install", "-q", "xgboost"]
    )
    from xgboost import XGBClassifier

print("All libraries imported successfully.")

# ================================================================
# 2. PROBLEM UNDERSTANDING
# ================================================================

display(Markdown("""
# Drinking Water Potability Prediction Using Machine Learning

## Problem Statement
The objective is to develop a supervised machine learning classification
system that predicts whether a water sample is potable using its
physicochemical properties.

## Project Objectives
1. Understand and explore the water potability dataset.
2. Identify missing values, duplicates and unusual observations.
3. Perform appropriate data cleaning and transformation.
4. Analyze and select informative features.
5. Develop four classification models.
6. Compare the models using multiple evaluation metrics.
7. Interpret the results and identify important features.

## KDD Approach
The Knowledge Discovery in Databases (KDD) process is followed through
data selection, preprocessing, transformation, data mining, evaluation
and knowledge presentation.
"""))

# ================================================================
# 3. DATA SELECTION AND LOADING
# ================================================================

DATA_PATH = "water_potability.csv"

if not os.path.exists(DATA_PATH):
    csv_files = [
        f for f in os.listdir(".")
        if f.lower().endswith(".csv")
    ]
    if len(csv_files) == 1:
        DATA_PATH = csv_files[0]
        print("Using detected CSV file:", DATA_PATH)
    else:
        raise FileNotFoundError(
            "CSV file not found. Download the confirmed Water Potability "
            "dataset, place its CSV in the notebook folder and set "
            "DATA_PATH to the correct filename. CSV files found: "
            + str(csv_files)
        )

df = pd.read_csv(DATA_PATH)

# Standardize column names for consistent processing.
df.columns = (
    df.columns.str.strip()
    .str.replace(" ", "_", regex=False)
)

# Accommodate common capitalization variations in the dataset.
column_mapping = {
    "ph": "ph",
    "PH": "ph",
    "Hardness": "Hardness",
    "Solids": "Solids",
    "Chloramines": "Chloramines",
    "Sulfate": "Sulfate",
    "Conductivity": "Conductivity",
    "Organic_carbon": "Organic_carbon",
    "Organic_Carbon": "Organic_carbon",
    "Trihalomethanes": "Trihalomethanes",
    "Turbidity": "Turbidity",
    "Potability": "Potability"
}
df = df.rename(columns=column_mapping)

print("\nDataset loaded successfully.")
print("Dataset shape:", df.shape)
print("Columns:", df.columns.tolist())

display(df.head())
display(df.sample(min(5, len(df)), random_state=RANDOM_STATE))

# Validate expected target and feature columns.
expected_features = [
    "ph", "Hardness", "Solids", "Chloramines", "Sulfate",
    "Conductivity", "Organic_carbon", "Trihalomethanes", "Turbidity"
]

if "Potability" not in df.columns:
    raise ValueError(
        "Potability target not found. Verify that the correct CSV was loaded."
    )

missing_columns = [
    col for col in expected_features if col not in df.columns
]
if missing_columns:
    raise ValueError(
        f"Expected water-quality features are missing: {missing_columns}. "
        "Verify the selected dataset and its column names."
    )

# Use the nine proposed physicochemical features.
X = df[expected_features].copy()
y = df["Potability"].copy()

# ================================================================
# 4. DATA UNDERSTANDING
# ================================================================

print("\n--- DATASET INFORMATION ---")
print("Number of records:", len(df))
print("Number of input features:", X.shape[1])
print("Target variable: Potability")
df.info()

print("\n--- DESCRIPTIVE STATISTICS ---")
display(X.describe().T.round(3))

print("\n--- DATA TYPES ---")
display(df.dtypes.to_frame("Data Type"))

print("\n--- MISSING VALUES ---")
missing_summary = pd.DataFrame({
    "Missing_Count": X.isnull().sum(),
    "Missing_Percentage": X.isnull().mean() * 100
})
display(missing_summary.round(2))

print("\n--- DUPLICATE RECORDS ---")
print("Duplicate rows:", df.duplicated().sum())

print("\n--- TARGET DISTRIBUTION ---")
display(y.value_counts().rename("Count").to_frame())
display(
    (y.value_counts(normalize=True) * 100)
    .rename("Percentage").to_frame().round(2)
)

if not set(y.dropna().unique()).issubset({0, 1}):
    raise ValueError(
        "Potability must be encoded as 0 and 1. Inspect the target values."
    )

if y.isnull().any():
    raise ValueError(
        "The target contains missing values. Investigate and handle them "
        "before modelling; do not impute the target automatically."
    )

# ================================================================
# 5. EXPLORATORY DATA ANALYSIS (EDA)
# ================================================================

# 5.1 Target distribution
plt.figure(figsize=(7, 4))
sns.countplot(data=df, x="Potability")
plt.title("Distribution of Water Potability")
plt.xlabel("Potability (0 = Not Potable, 1 = Potable)")
plt.ylabel("Number of Samples")
plt.tight_layout()
plt.show()

# 5.2 Feature distributions
df[expected_features].hist(
    figsize=(15, 11), bins=30, edgecolor="black"
)
plt.suptitle("Distribution of Physicochemical Features", y=1.02)
plt.tight_layout()
plt.show()

# 5.3 Boxplots to identify unusual observations
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
for ax, col in zip(axes.flatten(), expected_features):
    sns.boxplot(data=df, x=col, ax=ax)
    ax.set_title(f"Boxplot: {col}")
plt.tight_layout()
plt.show()

# 5.4 Correlation heatmap
plt.figure(figsize=(11, 8))
sns.heatmap(
    df[expected_features + ["Potability"]].corr(),
    annot=True, cmap="coolwarm", fmt=".2f", center=0
)
plt.title("Correlation Heatmap")
plt.tight_layout()
plt.show()

# 5.5 Feature distributions by target
fig, axes = plt.subplots(3, 3, figsize=(15, 12))
for ax, col in zip(axes.flatten(), expected_features):
    sns.boxplot(data=df, x="Potability", y=col, ax=ax)
    ax.set_title(f"{col} by Potability")
plt.tight_layout()
plt.show()

display(Markdown("""
### EDA Interpretation Guide
- Inspect the class distribution to determine whether the target is imbalanced.
- Use histograms to understand feature distributions and skewness.
- Use boxplots to identify potential outliers; an outlier is not automatically
  an error and should not be removed without justification.
- Use the correlation heatmap to identify linear relationships.
- Compare feature distributions across target classes to identify possible
  predictive patterns.
- Write conclusions based on the actual plots and statistics produced.
"""))

# ================================================================
# 6. DATA CLEANING
# ================================================================

# Remove exact duplicate rows.
before_duplicates = len(df)
df = df.drop_duplicates().reset_index(drop=True)
removed_duplicates = before_duplicates - len(df)

print("Duplicate rows removed:", removed_duplicates)

# Recreate X and y after duplicate removal.
X = df[expected_features].copy()
y = df["Potability"].copy()

# Convert feature columns to numeric where possible.
for col in expected_features:
    X[col] = pd.to_numeric(X[col], errors="coerce")

# Treat infinite numeric values as missing.
X = X.replace([np.inf, -np.inf], np.nan)

print("\nMissing values after numeric validation:")
display(X.isnull().sum().to_frame("Missing Values"))

# Do not impute before the train-test split.
# Median imputation will be fitted only on training data in the pipelines.

# Outlier detection using the IQR method.
outlier_rows = []
for col in expected_features:
    q1 = X[col].quantile(0.25)
    q3 = X[col].quantile(0.75)
    iqr = q3 - q1
    lower = q1 - 1.5 * iqr
    upper = q3 + 1.5 * iqr
    count = ((X[col] < lower) | (X[col] > upper)).sum()
    outlier_rows.append({
        "Feature": col,
        "IQR_Outlier_Count": int(count)
    })

outlier_summary = pd.DataFrame(outlier_rows)
display(outlier_summary)

print(
    "Outliers are reported for investigation and are not automatically "
    "removed, since they may represent valid water samples."
)

# ================================================================
# 7. DATA TRANSFORMATION AND TRAIN-TEST SPLIT
# ================================================================

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=RANDOM_STATE,
    stratify=y
)

print("Training set:", X_train.shape)
print("Testing set:", X_test.shape)
print("Training target proportions:")
display(y_train.value_counts(normalize=True).rename("Proportion").to_frame())

# Stratified splitting maintains similar class proportions.
# Imputation and scaling are placed inside pipelines to prevent leakage.

# ================================================================
# 8. FEATURE SELECTION / RELEVANCE ANALYSIS
# ================================================================

# Mutual information is estimated from training data only.
# Median imputation is fitted only on the training set for this analysis.
mi_imputer = SimpleImputer(strategy="median")
X_train_imputed = pd.DataFrame(
    mi_imputer.fit_transform(X_train),
    columns=expected_features,
    index=X_train.index
)

mi_scores = mutual_info_classif(
    X_train_imputed,
    y_train,
    random_state=RANDOM_STATE
)

feature_ranking = pd.DataFrame({
    "Feature": expected_features,
    "Mutual_Information": mi_scores
}).sort_values("Mutual_Information", ascending=False)

print("\nFeature relevance ranking:")
display(feature_ranking.round(4))

plt.figure(figsize=(9, 5))
sns.barplot(
    data=feature_ranking,
    x="Mutual_Information",
    y="Feature"
)
plt.title("Feature Relevance Using Mutual Information")
plt.tight_layout()
plt.show()

# Primary model comparison uses all nine proposed features.
# The top seven features are also evaluated in a separate experiment.
top_features = feature_ranking.head(
    min(7, len(expected_features))
)["Feature"].tolist()

print("\nSelected features for the additional experiment:")
print(top_features)

# ================================================================
# 9. DATA MINING: BUILD CLASSIFICATION MODELS
# ================================================================

# Calculate the class weight for XGBoost from training labels only.
negative_count = int((y_train == 0).sum())
positive_count = int((y_train == 1).sum())
scale_pos_weight = (
    negative_count / positive_count if positive_count > 0 else 1.0
)

models = {
    "Logistic Regression": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=2000,
            class_weight="balanced",
            random_state=RANDOM_STATE
        ))
    ]),

    "Decision Tree": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", DecisionTreeClassifier(
            max_depth=8,
            min_samples_leaf=5,
            class_weight="balanced",
            random_state=RANDOM_STATE
        ))
    ]),

    "Random Forest": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", RandomForestClassifier(
            n_estimators=300,
            min_samples_leaf=2,
            class_weight="balanced",
            random_state=RANDOM_STATE,
            n_jobs=-1
        ))
    ]),

    "XGBoost": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", XGBClassifier(
            n_estimators=300,
            max_depth=4,
            learning_rate=0.05,
            subsample=0.9,
            colsample_bytree=0.9,
            scale_pos_weight=scale_pos_weight,
            eval_metric="logloss",
            random_state=RANDOM_STATE,
            n_jobs=-1
        ))
    ])
}

# ================================================================
# 10. MODEL TRAINING AND EVALUATION
# ================================================================

results = []
trained_models = {}
test_probabilities = {}
test_predictions = {}

for name, model in models.items():
    print(f"\nTraining {name}...")

    model.fit(X_train, y_train)

    predictions = model.predict(X_test)
    probabilities = model.predict_proba(X_test)[:, 1]

    trained_models[name] = model
    test_predictions[name] = predictions
    test_probabilities[name] = probabilities

    results.append({
        "Model": name,
        "Accuracy": accuracy_score(y_test, predictions),
        "Precision": precision_score(
            y_test, predictions, zero_division=0
        ),
        "Recall": recall_score(
            y_test, predictions, zero_division=0
        ),
        "F1_Score": f1_score(
            y_test, predictions, zero_division=0
        ),
        "ROC_AUC": roc_auc_score(y_test, probabilities)
    })

results_df = pd.DataFrame(results).sort_values(
    by="F1_Score", ascending=False
).reset_index(drop=True)

print("\n--- TEST SET MODEL COMPARISON ---")
display(results_df.round(4))

# ================================================================
# 11. CROSS-VALIDATION
# ================================================================

# Five-fold stratified cross-validation on the training set.
# The held-out test set remains untouched during this step.
cv = StratifiedKFold(
    n_splits=5, shuffle=True, random_state=RANDOM_STATE
)

cv_results = []

for name, model in models.items():
    print(f"Cross-validating {name}...")

    scores = cross_val_score(
        model, X_train, y_train,
        cv=cv, scoring="f1", n_jobs=1
    )

    cv_results.append({
        "Model": name,
        "CV_F1_Mean": scores.mean(),
        "CV_F1_Std": scores.std()
    })

cv_df = pd.DataFrame(cv_results).sort_values(
    "CV_F1_Mean", ascending=False
).reset_index(drop=True)

print("\n--- 5-FOLD TRAINING CROSS-VALIDATION ---")
display(cv_df.round(4))

# ================================================================
# 12. VISUAL MODEL COMPARISON
# ================================================================

metrics_to_plot = [
    "Accuracy", "Precision", "Recall", "F1_Score", "ROC_AUC"
]

results_long = results_df.melt(
    id_vars="Model",
    value_vars=metrics_to_plot,
    var_name="Metric",
    value_name="Score"
)

plt.figure(figsize=(13, 6))
sns.barplot(
    data=results_long,
    x="Model", y="Score", hue="Metric"
)
plt.ylim(0, 1.05)
plt.title("Performance Comparison of Classification Models")
plt.xticks(rotation=15)
plt.tight_layout()
plt.show()

# ================================================================
# 13. SELECT THE BEST MODEL
# ================================================================

# Predefined selection criterion: highest held-out test F1-score.
# Review recall, ROC-AUC and class-specific errors before final selection.
best_name = results_df.iloc[0]["Model"]
best_model = trained_models[best_name]
best_pred = test_predictions[best_name]
best_prob = test_probabilities[best_name]

print("Best model by test F1-score:", best_name)
print("\nClassification Report:")
print(classification_report(
    y_test, best_pred,
    target_names=["Not Potable (0)", "Potable (1)"],
    zero_division=0
))

# Confusion matrix
cm = confusion_matrix(y_test, best_pred)
disp = ConfusionMatrixDisplay(
    confusion_matrix=cm,
    display_labels=["Not Potable", "Potable"]
)
disp.plot(cmap="Blues", values_format="d")
plt.title(f"Confusion Matrix - {best_name}")
plt.tight_layout()
plt.show()

# ROC curves for all models
plt.figure(figsize=(9, 7))
for name in models:
    RocCurveDisplay.from_predictions(
        y_test, test_probabilities[name],
        name=name, ax=plt.gca()
    )
plt.plot([0, 1], [0, 1], "k--", label="Random baseline")
plt.title("ROC Curve Comparison")
plt.tight_layout()
plt.show()

# ================================================================
# 14. FEATURE SELECTION EXPERIMENT
# ================================================================

# Compare the top-seven-feature experiment with the all-feature
# version of the same selected model family.
# Selection was based on training data only.

selected_models = {
    "Logistic Regression": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("scaler", StandardScaler()),
        ("classifier", LogisticRegression(
            max_iter=2000, class_weight="balanced",
            random_state=RANDOM_STATE
        ))
    ]),
    "Decision Tree": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", DecisionTreeClassifier(
            max_depth=8, min_samples_leaf=5,
            class_weight="balanced", random_state=RANDOM_STATE
        ))
    ]),
    "Random Forest": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", RandomForestClassifier(
            n_estimators=300, min_samples_leaf=2,
            class_weight="balanced", random_state=RANDOM_STATE,
            n_jobs=-1
        ))
    ]),
    "XGBoost": Pipeline([
        ("imputer", SimpleImputer(strategy="median")),
        ("classifier", XGBClassifier(
            n_estimators=300, max_depth=4, learning_rate=0.05,
            subsample=0.9, colsample_bytree=0.9,
            scale_pos_weight=scale_pos_weight,
            eval_metric="logloss", random_state=RANDOM_STATE,
            n_jobs=-1
        ))
    ])
}

selected_experiment_results = []

for name, model in selected_models.items():
    model.fit(X_train[top_features], y_train)
    pred = model.predict(X_test[top_features])
    prob = model.predict_proba(X_test[top_features])[:, 1]

    selected_experiment_results.append({
        "Model": name,
        "Feature_Set": "Top 7 features",
        "Accuracy": accuracy_score(y_test, pred),
        "Precision": precision_score(y_test, pred, zero_division=0),
        "Recall": recall_score(y_test, pred, zero_division=0),
        "F1_Score": f1_score(y_test, pred, zero_division=0),
        "ROC_AUC": roc_auc_score(y_test, prob)
    })

selected_experiment_df = pd.DataFrame(selected_experiment_results)

all_features_results = results_df.copy()
all_features_results["Feature_Set"] = "All 9 features"

feature_experiment_comparison = pd.concat(
    [
        all_features_results[
            ["Model", "Feature_Set", "Accuracy", "Precision",
             "Recall", "F1_Score", "ROC_AUC"]
        ],
        selected_experiment_df
    ],
    ignore_index=True
)

print("\n--- ALL FEATURES VS TOP-SEVEN FEATURES ---")
display(
    feature_experiment_comparison.sort_values(
        ["Model", "Feature_Set"]
    ).round(4)
)

# The feature-selection experiment is diagnostic.
# Do not claim it improved performance unless the results show that.

# ================================================================
# 15. MODEL INTERPRETATION
# ================================================================

# Feature importance for tree-based models, or coefficients for
# Logistic Regression. These indicate model associations, not causation.

classifier = best_model.named_steps["classifier"]

if hasattr(classifier, "feature_importances_"):
    importance_values = classifier.feature_importances_
    importance_label = "Feature Importance"
elif hasattr(classifier, "coef_"):
    importance_values = np.abs(classifier.coef_[0])
    importance_label = "Absolute Logistic Regression Coefficient"
else:
    importance_values = None
    importance_label = None

if importance_values is not None:
    importance_df = pd.DataFrame({
        "Feature": expected_features,
        "Importance": importance_values
    }).sort_values("Importance", ascending=False)

    print("\n--- FEATURE INTERPRETATION ---")
    display(importance_df.round(4))

    plt.figure(figsize=(9, 5))
    sns.barplot(
        data=importance_df,
        x="Importance", y="Feature"
    )
    plt.title(f"{importance_label} - {best_name}")
    plt.tight_layout()
    plt.show()

# ================================================================
# 16. SAMPLE PREDICTION
# ================================================================

sample = X_test.iloc[[0]]
actual_label = int(y_test.iloc[0])

sample_prediction = int(best_model.predict(sample)[0])
sample_probability = float(best_model.predict_proba(sample)[0, 1])

print("\n--- SAMPLE WATER POTABILITY PREDICTION ---")
display(sample)

print("Actual label:", actual_label)
print("Predicted label:", sample_prediction)
print(f"Estimated probability of Potability=1: "
      f"{sample_probability:.4f}")

if sample_prediction == 1:
    print("Model classification: Potable class predicted.")
else:
    print("Model classification: Non-potable class predicted.")

print(
    "Educational model output only; this prediction does not certify "
    "that a water sample is safe to drink."
)

# ================================================================
# 17. KNOWLEDGE PRESENTATION AND CONCLUSION
# ================================================================

best_row = results_df.loc[results_df["Model"] == best_name].iloc[0]
best_cv = cv_df.loc[cv_df["Model"] == best_name].iloc[0]

print("\n" + "=" * 65)
print("PROJECT RESULTS SUMMARY")
print("=" * 65)
print("Dataset:", DATA_PATH)
print("Final records after duplicate removal:", len(df))
print("Input features:", len(expected_features))
print("Training samples:", len(X_train))
print("Testing samples:", len(X_test))
print("Selected model by test F1-score:", best_name)
print(f"Accuracy:  {best_row['Accuracy']:.4f}")
print(f"Precision: {best_row['Precision']:.4f}")
print(f"Recall:    {best_row['Recall']:.4f}")
print(f"F1-score:  {best_row['F1_Score']:.4f}")
print(f"ROC-AUC:   {best_row['ROC_AUC']:.4f}")
print(f"Mean CV F1-score: {best_cv['CV_F1_Mean']:.4f}")
print("=" * 65)

display(Markdown(f"""
## Conclusion Template

The project developed four supervised machine learning classifiers for
water potability prediction using nine physicochemical features.

Based on the predefined test-set F1-score criterion, **{best_name}**
was the highest-ranked model in this experiment. Its measured test metrics
are shown in the results table above.

The final interpretation should discuss the model's precision, recall,
ROC-AUC, confusion matrix, cross-validation performance and important
features. These values describe performance on this dataset and do not
establish that the model is suitable for every water source.

### Limitations
- Dataset quality and representativeness affect the predictions.
- Missing-value imputation may introduce uncertainty.
- A random train-test split does not establish performance on new regions
  or future samples.
- Potability depends on microbiological and chemical safety criteria
  beyond the available features.
- Predictions must not replace laboratory testing or applicable
  drinking-water standards.
"""))

# ================================================================
# 18. SAVE RESULTS AND MODEL ARTIFACTS
# ================================================================

results_df.to_csv("model_comparison_results.csv", index=False)
cv_df.to_csv("cross_validation_results.csv", index=False)
feature_ranking.to_csv("feature_relevance_ranking.csv", index=False)
outlier_summary.to_csv("outlier_analysis.csv", index=False)
feature_experiment_comparison.to_csv(
    "feature_selection_comparison.csv", index=False
)

joblib.dump(best_model, "best_water_potability_model.joblib")
joblib.dump(expected_features, "model_feature_names.joblib")

print("\nSaved output files:")
print("- model_comparison_results.csv")
print("- cross_validation_results.csv")
print("- feature_relevance_ranking.csv")
print("- outlier_analysis.csv")
print("- feature_selection_comparison.csv")
print("- best_water_potability_model.joblib")
print("- model_feature_names.joblib")

print("\nEND OF KDD PREDICTIVE ANALYTICS PIPELINE")
```
